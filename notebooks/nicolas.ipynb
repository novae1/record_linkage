```python
import linktransformer as lt
import pandas as pd
import numpy as np
```

```python
file_path = '../data/restaurants.csv'
df = pd.read_csv(file_path)
df.head()
```

```python
test_df = df.drop(columns='cluster')

test_df.head()
```

```python
print([len(col) for col in df[['name', 'address', 'city', 'phone', 'category']].values.T])
```

```python
# columns = ['name', 'phone']
# the function all_pair_combos_evaluate doesn't accept lists as input, only single strings
# to fix this we'll concatenate the columns into a single string

columns = ['name', 'phone']

df['name_phone'] = df[columns].apply(lambda x: ' '.join(x.dropna()), axis=1)

lt.infer.all_pair_combos_evaluate(
    df=df,
    model='sentence-transformers/multi-qa-mpnet-base-dot-v1',
    left_on='name_phone',
    right_on='name_phone',
)
```
